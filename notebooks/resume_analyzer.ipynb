```python
import pdfplumber
skills = ['python', 'sql', 'machine learning', 'flask']
pdf = pdfplumber.open('resume.pdf')
text = ''
for page in pdf.pages:
    text += page.extract_text()
found_skills = []
for skill in skills:
    if skill.lower() in text.lower():
        found_skills.append(skill)

print('Skills Found:')
print(found_skills)
```
